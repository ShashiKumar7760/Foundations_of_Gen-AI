```python
# -*- coding: utf-8 -*-
"""Financial & Market Agent.ipynb

Automatically generated by Colab.

Original file is located at
    https://colab.research.google.com/drive/1a_InNcomP7xhva4S9gA3_If_qXl4ZvLX
"""



"""Problem Statement 2:


Develop LLM
agent with streamlit UX that return given country’s official currency, Real-time
exchange rate of 1 unit of that currency to USD, INR, GBP, and EUR, lists name
of major stock exchanges and indices in that country, current values of those
indices (real-time or recent), Google Maps pin for the stock exchange HQ


Framework: Lang Chain
/Phi Data or equivalent


LLM Engine: Gemini,
Lllama3, Mistral, Deep Seek or equivalent


Tools/Plugins: Yahoo
Finance, ExchangeRate-API, CurrencyAPI.com, Google Maps or Equivalent


Sample Prompt: Give me
currency and stock market details for Japan/India/US/South Korea/China/UK


Expected Output:


Name of the
official currency of the country


Exchange
rate of 1 JPY → USD, INR, GBP, EUR


Get
live/indexed values of indices (e.g., Nikkei225)


Display Google Maps pin for the
stock exchange HQ

"""

import streamlit as st
from phi.agent import Agent
from phi.model.google import Gemini
from phi.tools.yfinance import YFinanceTools
from phi.tools.duckduckgo import DuckDuckGo
import os

# Page Configuration
st.set_page_config(page_title="Global Finance Agent", page_icon="📈", layout="wide")

st.title("📈 Global Market & Currency Agent")
st.markdown("Get real-time currency, stock indices, and exchange locations.")

# Sidebar for API Key
api_key = st.sidebar.text_input("Enter Google Gemini API Key", type="password")

if not api_key:
    st.warning("Please enter your Google Gemini API Key in the sidebar to proceed.")
    st.stop()

os.environ["GOOGLE_API_KEY"] = "AIzaSyDs34RiywH0fF2Ksu8coHotJSO53mG7uLg"

# 1. Initialize the Agent
# We use YFinanceTools for stocks/currency and DuckDuckGo to find HQ addresses
finance_agent = Agent(
    name="Finance Agent",
    model=Gemini(id="gemini-2.5-flash"),
    tools=[
        YFinanceTools(stock_price=True, company_info=True, stock_fundamentals=True),
        DuckDuckGo()
    ],
    instructions=[
        "You are a financial analyst agent.",
        "Given a country name, perform the following steps:",
        "1. Identify the official currency.",
        "2. Find the exchange rate of 1 unit of that currency to USD, INR, GBP, and EUR using YFinance tickers (e.g., JPYUSD=X).",
        "3. List the major stock exchange name and its main index (e.g., Nikkei 225 for Japan).",
        "4. Get the CURRENT value of that index using YFinance.",
        "5. Search for the exact address of the Stock Exchange HQ.",
        "6. Construct a Google Maps Search URL for that address.",
        "Output Format:",
        "## [Country Name]",
        "**Official Currency:** [Name] ([Code])",
        "**Exchange Rates:**",
        "- 1 [Code] = [Value] USD",
        "- 1 [Code] = [Value] INR",
        "- 1 [Code] = [Value] GBP",
        "- 1 [Code] = [Value] EUR",
        "**Major Stock Index:**",
        "- Exchange: [Name]",
        "- Index: [Name] ([Ticker])",
        "- Value: [Current Price]",
        "**HQ Location:**",
        "[Address]",
        "[View on Google Maps](https://www.google.com/maps/search/?api=1&query=[Address_URL_Encoded])"
    ],
    show_tool_calls=True,
    markdown=True,
)

# 2. User Input
country = st.text_input("Enter Country Name", placeholder="e.g., Japan, India, USA")

# 3. Execution
if st.button("Get Financial Data 💰"):
    if country:
        prompt = f"Give me currency and stock market details for {country}"

        with st.spinner(f"Fetching real-time market data for {country}..."):
            output_container = st.empty()

            try:
                # Run agent
                response = finance_agent.run(prompt)

                # Display Markdown Output
                output_container.markdown(response.content)

                # Extract Google Maps Link (Basic parsing for UX enhancement)
                content = response.content
                if "maps/search" in content:
                    # Try to find a link to embed or highlight
                    st.success("Data retrieved successfully.")

            except Exception as e:
                st.error(f"Error fetching data: {e}")
    else:
        st.warning("Please enter a country name.")

# Optional: Sample Prompts
st.sidebar.markdown("### Sample Inputs")
st.sidebar.text("Japan")
st.sidebar.text("India")
st.sidebar.text("United Kingdom")
st.sidebar.text("USA")
```
