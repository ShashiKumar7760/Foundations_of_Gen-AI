```python
from tensorflow.keras.datasets import imdb
import numpy as np

# Load dataset
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=5000)
```

```python
def vectorize_sequences(sequences, dimension=5000):
    results = np.zeros((len(sequences), dimension))
    for i, sequence in enumerate(sequences):
        for index in sequence:
            if index < dimension:
                results[i][index] = 1.0
    return results
```

```python
x_train = vectorize_sequences(x_train)
x_test = vectorize_sequences(x_test)
```

```python
print(x_train.shape)
print(y_train.shape)
```

```python
# Initialize parameters
np.random.seed(42)
weights = np.random.randn(5000) * 0.01
bias = 0.0
```

```python
def sigmoid(z):
    return 1 / (1 + np.exp(-z))
```

```python
def compute_loss(y_true, y_pred):
    m = len(y_true)
    loss = - (1/m) * np.sum(
        y_true * np.log(y_pred + 1e-8) +
        (1 - y_true) * np.log(1 - y_pred + 1e-8)
    )
    return loss
```

```python
learning_rate = 0.1
epochs = 100

for epoch in range(epochs):

    # Forward pass
    z = np.dot(x_train, weights) + bias
    y_pred = sigmoid(z)

    # Compute loss
    loss = compute_loss(y_train, y_pred)

    # Backpropagation (gradients)
    m = len(y_train)
    dw = (1/m) * np.dot(x_train.T, (y_pred - y_train))
    db = (1/m) * np.sum(y_pred - y_train)

    # Update parameters
    weights -= learning_rate * dw
    bias -= learning_rate * db

    print(f"Epoch {epoch+1}, Loss: {loss:.4f}")
```

```python
# Training predictions
train_preds = sigmoid(np.dot(x_train, weights) + bias)
train_preds_binary = (train_preds >= 0.5).astype(int)

# Calculate accuracy
train_accuracy = np.mean(train_preds_binary == y_train)
print("Training Accuracy:", train_accuracy)
```

```python
# Test predictions
test_preds = sigmoid(np.dot(x_test, weights) + bias)
test_preds_binary = (test_preds >= 0.5).astype(int)

test_accuracy = np.mean(test_preds_binary == y_test)
print("Test Accuracy:", test_accuracy)
```

```python
# Load word index (IMPORTANT)
word_index = imdb.get_word_index()
```

```python
def review_to_vector(review):
    review = review.lower().split()
    vector = np.zeros(5000)

    for word in review:
        if word in word_index:
            index = word_index[word] + 3   # IMPORTANT FIX
            if index < 5000:
                vector[index] = 1

    return vector
```

```python
def predict_review(text):
    vector = review_to_vector(text)
    z = np.dot(vector, weights) + bias
    prob = sigmoid(z)

    sentiment = "Positive 😊" if prob >= 0.5 else "Negative 😞"

    print("Review:", text)
    print("Prediction:", sentiment)
    print("Confidence:", round(float(prob), 2))
```

```python
predict_review("This movie was absolutely amazing and fantastic")
predict_review("This was the worst movie")
```
