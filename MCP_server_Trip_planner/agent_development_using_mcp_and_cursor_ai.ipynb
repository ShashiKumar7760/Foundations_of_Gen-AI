```python
# -*- coding: utf-8 -*-
"""Agent development using MCP and Cursor AI.ipynb

Automatically generated by Colab.

Original file is located at
    https://colab.research.google.com/drive/11QS4hGwMHpQUSYdcu2p-SJYh60_p5H76

Problem Statement 1:  Agent development using MCP and Cursor AI

Develop LLM with streamlit UX agent that helps users plan a trip by combining LLM reasoning with real-time data (e.g., weather, flights, hotels).
Framework: Lang Chain /Phi Data or equivalent
LLM Engine: Gemini, Lllama3, Mistral, Deep Seek or equivalent
Tools/Plugins: Weather API (e.g., Open Weather), Flight API (e.g. Skyscanner), Use Hotel API, Google Places or equivalent
Sample Prompts: Plan a 3-day trip to Tokyo in May, Plan a 2-day trip to Udaipur in May
Expected Output:
1 paragraph about city’s cultural & historic significance, Current Weather and Weather forecast during trip
Travel dates, Flight & Hotel Options, Trip plan
"""

#pip install streamlit phidata google-generativeai duckduckgo-search yfinance packaging requests

import streamlit as st
from phi.agent import Agent
from phi.model.google import Gemini
from phi.tools.duckduckgo import DuckDuckGo
import os

# Page Configuration
st.set_page_config(page_title="AI Travel Planner", page_icon="✈️", layout="wide")

st.title("✈️ AI Trip Planner Agent")
st.markdown("Plan your perfect trip with real-time data using Gemini & PhiData.")

# Sidebar for API Key
api_key = st.sidebar.text_input("Enter Google Gemini API Key", type="password")

if not api_key:
    st.warning("Please enter your Google Gemini API Key in the sidebar to proceed.")
    st.stop()

os.environ["GOOGLE_API_KEY"] = api_key

# 1. Initialize the Agent with Tools
# We use DuckDuckGo for Search (Weather, Flights, Hotels, Culture)
travel_agent = Agent(
    name="Travel Planner",
    model=Gemini(id="gemini-2.5-flash"),
    tools=[DuckDuckGo()],
    instructions=[
        "You are a helpful travel assistant.",
        "When asked to plan a trip, structure your answer exactly as follows:",
        "1. **Cultural & Historic Significance**: A concise paragraph about the city.",
        "2. **Current Weather & Forecast**: Find current weather and historical/forecast trends for the travel month.",
        "3. **Travel Logistics**: Suggest specific travel dates, 3 flight options (airlines/approx prices), and 3 hotel options (names/ratings/approx prices).",
        "4. **Trip Itinerary**: A day-by-day plan based on the duration requested.",
        "Always use the search tool to find real, up-to-date information for prices and weather."
    ],
    show_tool_calls=True,
    markdown=True,
)

# 2. User Inputs
col1, col2, col3 = st.columns(3)
with col1:
    destination = st.text_input("Destination", placeholder="e.g., Tokyo")
with col2:
    month = st.text_input("Travel Month", placeholder="e.g., May")
with col3:
    duration = st.text_input("Duration", placeholder="e.g., 3 days")

# 3. Execution
if st.button("Plan My Trip 🌍"):
    if destination and month and duration:
        prompt = f"Plan a {duration} trip to {destination} in {month}."

        with st.spinner("Agent is researching weather, flights, and hotels..."):
            # Create a container for the output
            response_container = st.empty()

            # Run the agent and stream the response
            try:
                response = travel_agent.run(prompt)
                response_container.markdown(response.content)
            except Exception as e:
                st.error(f"An error occurred: {e}")
    else:
        st.error("Please fill in all fields.")

# Optional: Display Example Prompts
st.sidebar.markdown("### Sample Prompts")
st.sidebar.info("- Plan a 3-day trip to Tokyo in May\n- Plan a 2-day trip to Udaipur in December")
```
